## About the Data

This dataset contains compiled occurrence records for prey items of listed (endangered) terrestrial vertebrate species within central California drylands, along with associated environmental data. It is a synthesis of records from the Global Biodiversity Information Facility (GBIF) and peer-reviewed literature, spanning observations from 1945 to 2022.

**Citation:** King, R., Braun, J., Westphal, M., & Lortie, C.J. (2023). Compiled occurrence records for prey items of listed species found in California drylands with associated environmental data. Knowledge Network for Biocomplexity. doi:10.5063/F1VM49RH.

**Date accessed:** October 2, 2026

**Link:** [https://knb.ecoinformatics.org/view/doi:10.5063/F1VM49RH](https://knb.ecoinformatics.org/view/doi:10.5063/F1VM49RH)

## Archive Exploration

**1. What is this data about?**

It's an open dataset summarizing all the reported diet and prey items for all endangered, terrestrial dryland species listed in central California. Specifically, it contains compiled occurrence records for prey species of 11 listed species found in California drylands, sourced from the Global Biodiversity Information Facility (GBIF).

**2. Is this data collected in situ by the authors or is it a synthesis?**

It's a synthesis. The search terms included the Latin binomial for each listed species and the terms "diet" and "prey" listed anywhere within publications. The authors compiled records from existing literature and GBIF rather than collecting new field observations themselves.

**3. During what time frame were the observations collected?**

The compiled observations span the time period 1945–2022.

**4. Does this dataset come with an associated metadata file?**

Yes. The archive includes an `.xml` document in EML format (Ecological Metadata Language), which is a machine-readable file that has metadata about the whole dataset. There is also a metadata CSV file with column descriptions for the data files.

**5. Does the dataset contain sensitive data?**

Yes — the dataset includes precise geographic occurrence locations for prey species associated with endangered and listed vertebrate species. Location data for endangered species is generally considered sensitive because it could potentially be misused (for example, for poaching or habitat disturbance). However, in this case the authors chose to make it openly available through KNB to support conservation planning.
